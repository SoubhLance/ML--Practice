## Magic Module

#### the "__init__" Module


```python
class Student():
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
Student('Soubhik',24)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}, {self.age}"
```

```python
s = Student("Soubhik", 24)
print(s)
```
