### ❓ Question

Fill in the blank so that the code correctly appends the **antonym word**.

> ✏️ Hint: `l.antonyms()` returns a list of lemma objects.


````python
from nltk.corpus import wordnet

synonyms = []
antonyms = []

for syn in wordnet.synsets("good"):
    for l in syn.lemmas():
        synonyms.append(l.name())
        if l.antonyms():
           antonyms.append(l.antonyms()[0].name())

print(set(synonyms))
print(set(antonyms))
````

````python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "Mary is going to the market"
words = word_tokenize(text)

filtered = [w for w in words if w.lower() not in stopwords.words('english')]

print(filtered)
````
