## What is a corpus in NLP?

### A corpus in NLP is a large collection of text data used for training and evaluating language models and other NLP algorithms.

It serves as a representative sample of language usage and can include various types of text, such as books, articles, social media posts, and more.

A corpus is essential for tasks like:

* language modeling
* sentiment analysis
* machine translation

as it provides the necessary data for learning patterns and making predictions in natural language processing applications.

---

#### **Corpus = text ka bada collection (data set)**

---

#### **Yani bohot saare:**

* sentences
* documents
* articles
* ya words ka data

---

#### **jo NLP me training / analysis ke liye use hota hai.**


```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('stopwords')

```

```python
words = ['I', 'love', 'learning', 'natural', 'language', 'processing']
filtered_words = [w for w in words if w.lower() not in stopwords.words('english')]
print(filtered_words)
```

## Stemming

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()
print(ps.stem("playing"))
```

```python
words = ["run", "running", "runner", "ran"]
stemmed_words = [ps.stem(word) for word in words]
print(stemmed_words)    
```

```python
s = "Hello How are you doing today?"
s = s.lower()
tokens = word_tokenize(s)
data = [w for w in tokens if w.lower() not in stopwords.words('english')]
data = [ps.stem(w) for w in data]
print(data)
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["playing", "played", "plays"]
result = [ps.stem(w) for w in words]

print(result)
```

## Lemmatization (Smart Root)

Lemmatization reduces a word to its **meaningful base form (lemma)** using vocabulary and context.

### 🔑 POS Tags (Important)

* **'n'** → Noun *(default)*
* **'v'** → Verb
* **'a'** → Adjective
* **'r'** → Adverb

> 💡 Tip: Always provide the correct POS tag for accurate results.


```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python

lem = WordNetLemmatizer()
print(lem.lemmatize("running"))
```

```python

lem = WordNetLemmatizer()
print(lem.lemmatize("better"))
```

```python
from nltk.stem import WordNetLemmatizer

lem = WordNetLemmatizer()

print(lem.lemmatize("running", pos='v'))   # verb
print(lem.lemmatize("better", pos='a'))    # adjective
```
